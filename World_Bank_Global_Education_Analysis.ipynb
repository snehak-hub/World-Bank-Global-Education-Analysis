## Project Summary

The** World Bank Education Statistics (EdStats) Analysis** project focuses on exploring and analyzing the World Bank EdStats All Indicator Query dataset, which contains over **4,000 internationally comparable education indicators** collected from more than **200 countries and regions**. The dataset provides a comprehensive view of global education by covering key areas such as school enrollment, literacy rates, educational attainment, teacher statistics, government expenditure on education, population demographics, and learning outcomes across different education levels, from pre-primary to tertiary education. The primary objective of this project is to perform **Exploratory Data Analysis (EDA)** to understand global education trends, compare the educational performance of different countries, identify similarities and differences among regions, and examine the relationships between important education indicators such as literacy, enrollment, and public spending. Since the raw dataset contains a large number of indicators and missing values, the first step involved cleaning and preprocessing the data by handling null values, removing unnecessary columns, selecting relevant indicators, and transforming the data into a suitable format for analysis. **Python** was used as the primary programming language because of its powerful data analysis capabilities. **Pandas** was used for data manipulation, filtering, grouping, and aggregation, while **NumPy** supported efficient numerical computations and statistical operations. The cleaned data was then explored using descriptive statistics and various analytical techniques to understand distributions, identify trends, and detect patterns that could provide meaningful insights into the global education system.

To effectively communicate the findings, multiple visualization techniques were created using **Matplotlib** and **Seaborn**, including **bar charts, line charts, histograms, scatter plots, heatmaps, and box plots**. These visualizations helped compare literacy rates, enrollment levels, education expenditure, and other indicators across countries and regions, making complex data easier to understand. The analysis highlighted countries with high and low educational performance, revealed regional disparities in access to education, and showed how government investment in education is associated with improvements in literacy and enrollment in many cases. It also identified trends over time where historical data was available, enabling a better understanding of educational progress and challenges faced by different nations. Through this project, valuable insights were generated that can assist policymakers, researchers, educators, and organizations in understanding global education patterns and supporting evidence-based decision-making. Beyond the analytical results, the project strengthened practical skills in **data cleaning, preprocessing, exploratory data analysis (EDA), statistical analysis, feature selection, data visualization, and insight generation** using real-world data. It also improved proficiency in working with large datasets, interpreting statistical outputs, and presenting findings in a clear and meaningful manner. Overall, this project demonstrates how data analytics can be applied to solve real-world problems by transforming complex educational data into actionable insights, while showcasing the effective use of **Python, Pandas, NumPy, Matplotlib, Seaborn, and Jupyter Notebook** for end-to-end data analysis and visualization.

#### Chart - 4-1. Education expenditure by region (Bar chart)

````python
education = EData[
    EData['Indicator Name'].str.contains(
        'expenditure on education',
        case=False,
        na=False
    )
].copy()

education['2000'] = pd.to_numeric(
    education['2000'],
    errors='coerce'
)

region_education = (
    education.dropna(subset=['Region_x', '2000'])
    .groupby('Region_x')['2000']
    .mean()
    .sort_values()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=region_education.values,
    y=region_education.index
)

plt.title("Average Education Expenditure by Region - 2000")
plt.xlabel("Education Expenditure")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
````

#### Chart - 12-Pair Plot with Region

````python
# Chart - 12 visualization code
pair_data = EData[['Region_x', '2015', '2020']].copy()

pair_data[['2015', '2020']] = pair_data[['2015', '2020']].apply(
    pd.to_numeric, errors='coerce'
)

pair_data = pair_data.dropna()

# Limit rows for faster plotting
pair_data = pair_data.sample(500, random_state=42)

sns.pairplot(
    pair_data,
    vars=['2015', '2020'],
    hue='Region_x'
)

plt.show()
````
